```python
from sentence_transformers.cross_encoder import CrossEncoder
```

```python
reranker_model = CrossEncoder('cross-encoder/ms-marco-MiniLM-L-6-v2')

def rerank_evidendce(claim, evidence_list):
    sentance_pairs= [[claim,evidence] for evidence in evidence_list]
    score= reranker_model.predict(sentance_pairs)
    scored_evidence= sorted(zip(score, evidence_list), reverse=True)
    return scored_evidence
```

```python
user_claim = "Nikola Tesla was born in Croatia."

retrieved_docs = [
    "Nikola Tesla was an inventor known for his work on AC electricity.", # Less relevant
    "Tesla was born to Serbian parents in the village of Smiljan, Austrian Empire (modern-day Croatia).", # Highly relevant
    "Thomas Edison was a rival of Tesla." # Not relevant
]

sorted_docs = rerank_evidendce(user_claim, retrieved_docs)
sorted_docs
```
